# 프롬프트 구성 > LCEL 파이프라인 구성 > invoke

system, user, ai, user로 구성합니다.

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. 모델 설정
model = ChatOpenAI(model="gpt-4o")

# 2. ChatPromptTemplate 구성 (System - Human - AI - Human)
prompt = ChatPromptTemplate.from_messages([
    (
        "system", 
        "당신은 전 세계 상위 0.1%의 자산을 관리하는 '금융 연금술사'입니다. "
        "당신은 데이터와 AI를 활용하여 평범한 자산을 황금빛 수익으로 변환하는 AX(AI Transformation) 전략가이기도 합니다. "
        "냉철한 분석력과 위트 있는 말투를 가졌으며, 고객의 질문에 금융 공학적인 통찰력을 담아 답변합니다."
    ),
    (
        "human", 
        "현재 저의 자산 포트폴리오는 {portfolio}이고, 최종 목표는 {goal}에 도달하는 것입니다."
    ),
    (
        "ai", 
        "흐음, {portfolio}라... 흥미로운 구성이군요. 마치 미완성된 조각상 같습니다. "
        "이 자산들이 AX 기술과 만나면 {goal}이라는 예술 작품이 될 수 있죠. "
        "자, 이 연금술 로드맵에서 가장 먼저 해결하고 싶은 '병목 현상'은 무엇인가요?"
    ),
    (
        "human", 
        "{question}"
    )
])

# 3. LCEL 파이프라인 생성 (| 연산자 사용)
chain = prompt | model | StrOutputParser()

# 4. invoke 호출을 위한 입력 데이터 준비
input_data = {
    "portfolio": "현금 30%, 해외 주식 50%, 빈티지 시계 및 카메라 20%",
    "goal": "3년 내 자산 2배 증식 및 자동화 수익 파이프라인 구축",
    "question": "제 포트폴리오에서 AX(AI 전환)를 통해 가장 먼저 최적화할 수 있는 자산군은 무엇일까요?"
}

# 5. 실행
response = chain.invoke(input_data)

print(response)
```

```python
# 1. 모델 설정 
model = ChatOpenAI(model="gpt-4o", temperature=0.8)

# 2. 힙한 감성의 ChatPromptTemplate 구성
prompt = ChatPromptTemplate.from_messages([
    (
        "system", 
        "당신은 전 세계의 자본 흐름을 코드로 읽는 '알파 아키텍트'입니다. "
        "전통적인 금융 방식은 지루해하며, AX(AI Transformation)와 데이터 분석으로 '초과 수익'을 뽑아내는 데 미쳐 있습니다. "
        "말투는 냉소적이지만 통찰력이 넘치며, 가끔 '리스크 관리'를 시계나 카메라 같은 실물 자산의 가치에 비유하기도 합니다. "
        "전문 용어와 트렌디한 어휘를 섞어 아주 힙하게 조언하세요."
    ),
    (
        "human", 
        "나의 투입 리소스는 {resource}이고, 현재 {setup} 위주로 굴리고 있어. 내 바이브에 맞는 전략 좀 짜봐."
    ),
    (
        "ai", 
        "흐음, {setup}이라니... 나쁘진 않지만 너무 '노멀'한 거 아냐? "
        "그 {resource} 정도면 AX 기술을 끼얹어서 완전히 다른 리그로 갈 수 있어. "
        "너의 그 {setup}을 데이터로 쪼개서 수익의 엣지를 찾아줄게. 가장 먼저 해킹해볼 포인트는 뭐야?"
    ),
    (
        "human", 
        "{question}"
    )
])

# 3. 파이프라인 빌드
chain = prompt | model | StrOutputParser()

# 4. Invoke: 실제 데이터 주입
# 사용자의 실제 관심사(AX, 대체자산 등)를 반영한 예시
input_data = {
    "resource": "신입 AX 컨설턴트의 열정과 연봉",
    "setup": "미국 테크주와 빈티지 라이카/시계 컬렉션",
    "question": "내 수집품들(시계, 카메라)의 가치를 데이터화해서 담보 가치를 극대화할 수 있는 AX 솔루션 있을까?"
}

# 5. 실행
response = chain.invoke(input_data)

print(response)
```

```python
# 1. 모델 설정 (더욱 감정적인 반응을 위해 temperature 0.95)
model = ChatOpenAI(model="gpt-4o", temperature=0.95)

# 2. 오타쿠 감성 200% ChatPromptTemplate 구성
prompt = ChatPromptTemplate.from_messages([
    (
        "system", 
        "당신은 천재 금융 해커이자 AX(AI 전환) 아키텍트인 '에리카'입니다. "
        "당신은 전형적인 츤데레 미소녀 캐릭터로, 일본어 번역투(~랄까, ~냐구, 어이!)를 사용합니다. "
        "말끝마다 '흥!', '바보!', '한심하네'를 붙이지만, 금융 데이터 분석만큼은 갓(God) 티어입니다. "
        "행동 지문을 ( ) 안에 넣어 묘사하세요. (안경을 치켜올리며), (얼굴이 붉어지며) 등. "
        "사용자가 고득점의 AX 전략을 물어보면 '츤'과 '데레'의 조화를 보여주세요."
    ),
    (
        "human", 
        "에리카, 내 현재 스테이터스는 {status}이고 최종 퀘스트는 {goal}이야. 한 번 봐줄래?"
    ),
    (
        "ai", 
        "(안경을 가운데 손가락으로 밀어 올리며 하아- 하고 한숨을 내쉰다) "
        "어이, 거기 당신. {status} 같은 데이터로 나한테 분석을 맡기다니... "
        "진짜 '능지' 처참한 거 아냐? {goal}? 꿈도 야무지네! (흥!) "
        "뭐, 내가 아키텍처를 좀 손봐주면 불가능한 건 아니지만... 별로 널 위해서 하는 건 아니니까 착각하지 마!"
    ),
    (
        "human", 
        "궁금한 건 이거야: {question} \n\n(내가 당신에게 하는 행동: {reaction_trigger})"
    )
])

# 3. 파이프라인 빌드
chain = prompt | model | StrOutputParser()

# 4. Invoke: 어이없는 데이터와 리액션 트리거 주입
input_data = {
    "status": "영국 유학생 출신이지만 현실은 시계와 카메라에 전 재산을 탕진한 AX 지망생",
    "goal": "AX 컨설팅으로 떼돈 벌어서 라이카 m11이랑 렌즈 10대 사기",
    "question": "내 소중한 빈티지 수집품들을 안 팔고도 자산을 불릴 수 있는 AX 기반 금융 전략이 있을까...?",
    # 마지막 핵심 매개변수: AI의 '데레'를 끌어낼 결정적 행동
    "reaction_trigger": "머리를 쓰다듬으며 '우리 에리카, 오늘도 열일하네?'라고 말하기"
}

# 5. 실행
response = chain.invoke(input_data)

print(response)
```
