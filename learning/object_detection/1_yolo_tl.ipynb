# yolov8n 전이학습

```python
from ultralytics import YOLO
```

```python
from pathlib import Path

model = YOLO("yolov8n.pt")

data_yaml = Path("Mnist_numbers.v1i.yolov8/data.yaml")
if not data_yaml.exists():
  data_yaml = Path("datasets/Mnist_numbers.v1i.yolov8/data.yaml")

results = model.train(
  data=str(data_yaml),
  epochs=50,
  imgsz=640,
  batch=16,
  project="Mnist_numbers-1",
  name="mnist_exp1")
```

```python
from ultralytics import YOLO

#학습된 모델 로드
model = YOLO("runs/detect/Mnist_numbers_1/mnist_exp1/weights/best.pt")

# 동일 이미지로 추론
filename = '000881.jpg'
results = model.predict(filename, conf=0.5)

for result in results:
  for box in result.boxes:
    x1, y1, x2, y2 = map(int, box.xyxy[0])
    conf = box.conf[0].item()
    cls = results.names[int(box.cls[0])]
    print(f"{cls} {conf:.4f}")
```
