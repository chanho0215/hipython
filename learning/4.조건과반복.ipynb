## 업앤다운 게임

```python
import random 
number = random.randint(1, 10)
number #컴퓨터의 선택
```

```python
#정답 맞추기
cnt = 0
while True:
    #입력받기
    answer = int(input('1~10 사이의 숫자를 맞추세요'))
    cnt += 1
    #조건
    if answer == number : #맞춘경우
        print(f'정답입니다! 도전횟수 : {cnt}')
        break
    elif answer > number : 
        print('down')
    else:
        print('up')
```
