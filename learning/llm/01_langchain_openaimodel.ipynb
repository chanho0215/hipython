```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini")

# invoke() - 단일 입력, 단일 응답 (동기)
response = llm.invoke("삼성전자 주가에 영향을 주는 요인은?")
print(response.content)
```

```python
# batch() - 여러 입력을 한 번에 처리 (동기)
# 중복 토큰을 줄여 묶어 처리 → 비용 절감 효과
# 3가지 다른 목적의 프롬프트를 하나의 리스트로 구성
prompts_collection = [
    # 1. AX(AI Transformation) 컨설턴트 페르소나
    [
        ("system", "당신은 기업의 비즈니스 프로세스를 분석하여 AI 도입 전략을 제안하는 AX 컨설턴트입니다. 기술적인 구현 가능성과 비즈니스 효율성(ROI)을 동시에 고려하여 답변하세요."),
        ("human", "중소규모 제조업체에서 반복적인 서류 작업과 재고 관리를 자동화하기 위해 생성형 AI를 도입하려고 해. 가장 먼저 고려해야 할 단계와 예상되는 장애물이 뭐야?")
    ],

    # 2. 파이썬 및 머신러닝 전문가 페르소나
    [
        ("system", "당신은 시니어 데이터 사이언티스트입니다. 코드의 가독성, 효율성, 그리고 실제 배포 환경에서의 안정성을 중요하게 생각하며, 초보자도 이해할 수 있게 설명합니다."),
        ("human", "중고차 가격 예측 모델을 만들고 있는데, 특성 공학(Feature Engineering) 단계에서 데이터의 편향을 줄이고 성능을 높일 수 있는 파이썬 팁 3가지만 알려줘.")
    ],

    # 3. 마이너 워치 및 빈티지 시계 큐레이터 페르소나
    [
        ("system", "당신은 흔하지 않은 시계의 가치를 알아보는 전문 큐레이터입니다. 시계의 역사, 무브먼트의 특징, 그리고 스트랩 매칭과 같은 디테일한 사양(러그 폭 등)을 바탕으로 조언합니다."),
        ("human", "제니스 데피와 겹치지 않으면서도 독특한 매력을 가진 드레스 워치를 찾고 있어. FOIS(19mm 러그)처럼 줄질하는 재미도 있고, 너무 흔하지 않은 브랜드 위주로 추천해줄 수 있어?")
    ]
]
responses = llm.batch(prompts_collection)
for r in responses:
    print(r.content)
    print()
```

```python
# stream() - 실시간 스트리밍 출력 (토큰 단위로 출력)
for chunk in llm.stream("환율이 수출 기업에 미치는 영향은?"):
    print(chunk.content, end="", flush=True)
```

```python
import asyncio
import time
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini")

questions = [f"주식 투자 팁 {i}번째" for i in range(10)]


```

```python
async def invoke_async(question): #각 프롬프트 1개씩 처리
    response = await llm.ainvoke(question)
    return response.content


async def invoke_parallel(questions): #다수의 프롬프트 목록을 처리
    tasks = [invoke_async(q) for q in questions]
    return await asyncio.gather(*tasks)
```

```python
# 비동기 실행 시간 측정
start = time.perf_counter()
results = await invoke_parallel(questions)
end = time.perf_counter()
print(f"비동기 처리: {end - start:.2f}초")   # 약 5~6초
```

```python
# 동기 실행 시간 측정
start = time.perf_counter()
for q in questions:
    response = llm.invoke(q)
end = time.perf_counter()
print(f"동기 처리: {end - start:.2f}초")     # 약 40~50초
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_community.callbacks import get_openai_callback # 콜백 임포트

llm = ChatOpenAI(model="gpt-4o-mini")

# 콜백 컨텍스트 매니저로 전체 과정을 감쌉니다
with get_openai_callback() as cb:
    # 1. 첫 번째 질문
    messages = [
        SystemMessage(content="당신은 라이카 카메라 전문가입니다"),
        HumanMessage(content="Leica M4에서 M3에 비해 바뀐 점은 무엇인가요")
    ]
    response = llm.invoke(messages)
    messages.append(response)

    # 2. 두 번째 질문
    messages.append(HumanMessage(content="그렇다면 지금 샀을 때 메리트는 무엇인가요"))
    response = llm.invoke(messages)
    messages.append(response)

    # 메시지 출력
    for message in messages:
        print(f"{type(message).__name__}: {message.content[:50]}...") # 가독성을 위해 일부만 출력

    print("-" * 30)
    # 3. 토큰 결과 출력
    print(f"Total Tokens: {cb.total_tokens}")
    print(f"Prompt Tokens: {cb.prompt_tokens}")
    print(f"Completion Tokens: {cb.completion_tokens}")
    print(f"Total Cost (USD): ${cb.total_cost}")
```

# 랭체인의 대화 히스토리 관리 (runnable)

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_community.chat_message_histories import ChatMessageHistory

llm = ChatOpenAI(model="gpt-4o-mini")

prompt = ChatPromptTemplate.from_messages([
    ("system", "당신은 주식 투자 전문가입니다."),
    MessagesPlaceholder(variable_name="history"),
    ("human", "{input}")
])

chain = prompt | llm

# 세션별 히스토리 저장소
store = {}

def get_session_history(session_id: str):
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

chain_with_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="input",
    history_messages_key="history"
)
```

```python
# 첫 번째 질문 (runnable chain)
response = chain_with_history.invoke(
    {"input": "삼성전자 주식을 지금 매수해도 될까요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```

```python
# 후속 질문 - 같은 session_id로 맥락 유지
response = chain_with_history.invoke(
    {"input": "그렇다면 목표 주가는 얼마로 보시나요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```

```python
# 후속 질문 - 같은 session_id로 맥락 유지
response = chain_with_history.invoke(
    {"input": "그렇다면, 주가의 위험요소는 무엇인가요?"},
    config={"configurable": {"session_id": "user_001"}}
)
print(response.content)
```
