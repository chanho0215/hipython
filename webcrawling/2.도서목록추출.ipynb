```python
import requests
from bs4 import BeautifulSoup

headers = {"User-Agent": "Mozilla/5.0"}
book_list = []

for page in range(1, 4):
    url = f"https://www.yes24.com/product/category/bestseller?categoryNumber=001&pageNumber={page}&pageSize=24"
    res = requests.get(url, headers=headers)
    soup = BeautifulSoup(res.text, "html.parser")

    # ✅ 책 단위 기준: 상품 상세 링크
    goods_links = soup.select('a[href*="/product/goods/"]')

    # 같은 책 링크가 여러 번 잡힐 수 있어서 set으로 중복 제거
    seen = set()

    for i, a in enumerate(goods_links):
        href = a.get("href", "")
        if "/product/goods/" not in href:
            continue
        if href in seen:
            continue
        seen.add(href)

        # ✅ "책 카드" 영역으로 올라가기 (가능한 가까운 li/div)
        card = a.find_parent(["li", "div"]) or a

        title = a.get_text(strip=True)
        author = card.select_one("span.authPub.info_auth a")
        pub = card.select_one("span.authPub.info_pub a")
        date = card.select_one("span.authPub.info_date")
        price = card.select_one("strong.txt_num")

        # ✅ 순위: 페이지 전체 순위로 계산 (HTML rank 없어도 됨)
        rank = (page - 1) * 24 + (len(book_list) % 24) + 1

        if title and price:
            book_list.append({
                "rank": rank,
                "title": title,
                "author": author.get_text(strip=True) if author else "저자 없음",
                "publisher": pub.get_text(strip=True) if pub else "출판사 없음",
                "publish_date": date.get_text(strip=True) if date else "출판일 없음",
                "price": price.get_text(strip=True) if price else ""
            })

print(len(book_list))
print(book_list[:3])
```
