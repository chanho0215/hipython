```python
import re
import time
import pandas as pd
import requests
from bs4 import BeautifulSoup

BASE = "https://finance.naver.com"

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0.0.0 Safari/537.36"
    )
}

def _get_redirect_url(soup: BeautifulSoup) -> str | None:
    """
    finance.naver.com 상세 페이지의 script에서 redirect URL을 뽑아냄.
    예: location.href='https://....'
    """
    script = soup.find("script")
    if not script or not script.get_text(strip=True):
        return None

    # 기존 split("'")[1] 방식보다 구조 변화에 덜 취약하게 정규식 사용
    m = re.search(r"['\"](https?://[^'\"]+)['\"]", script.get_text())
    return m.group(1) if m else None

def _fetch_article_text(session: requests.Session, url: str, timeout: int = 10) -> str | None:
    """
    최종 뉴스 URL에서 본문 텍스트를 파싱.
    """
    r = session.get(url, timeout=timeout)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "lxml")

    body = soup.select_one("div.newsct_article._article_body")
    if not body:
        return None

    return body.get_text(" ", strip=True)

def _resolve_final_news_url(session: requests.Session, url: str, timeout: int = 10) -> str | None:
    """
    finance.naver.com 기사 링크(중간 페이지) -> script 리다이렉트로 최종 뉴스 URL 얻기
    """
    r = session.get(url, timeout=timeout)
    r.raise_for_status()
    soup = BeautifulSoup(r.text, "lxml")
    return _get_redirect_url(soup)

def get_news_items(html: BeautifulSoup, sleep_sec: float = 0.2, timeout: int = 10) -> pd.DataFrame:
    """
    dd.articleSubject 리스트에서 기사 제목/최종URL/본문을 가져와 DataFrame으로 반환.
    """
    items = html.find_all("dd", {"class": "articleSubject"})

    rows = []
    with requests.Session() as session:
        session.headers.update(HEADERS)

        for it in items:
            try:
                a = it.find("a")
                if not a:
                    continue

                title = it.get_text(strip=True)
                mid_url = BASE + a.get("href", "")

                final_url = _resolve_final_news_url(session, mid_url, timeout=timeout)
                if not final_url:
                    rows.append({"기사제목": title, "본문url": None, "기사본문": None})
                    continue

                article = _fetch_article_text(session, final_url, timeout=timeout)

                rows.append({"기사제목": title, "본문url": final_url, "기사본문": article})

                if sleep_sec:
                    time.sleep(sleep_sec)

            except requests.RequestException:
                # 요청/타임아웃/HTTP 에러 등은 해당 기사만 None 처리하고 계속
                rows.append({"기사제목": it.get_text(strip=True), "본문url": None, "기사본문": None})
            except Exception:
                # HTML 구조 변화 등 예기치 못한 에러도 일단 기사 단위로 방어
                rows.append({"기사제목": it.get_text(strip=True), "본문url": None, "기사본문": None})

    df = pd.DataFrame(rows).drop_duplicates(subset=["기사제목", "본문url"], keep="first")
    return df
```

```python
import datetime
import time
import pandas as pd
import requests
from bs4 import BeautifulSoup

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0.0.0 Safari/537.36"
    )
}

def scrape_naver_finance_news(days: int = 3,
                              sleep_sec: float = 1.0,
                              timeout: int = 10,
                              out_csv: str = "article.csv",
                              verbose: bool = False) -> pd.DataFrame:
    """
    오늘 포함 days일치 네이버금융 메인뉴스를 수집해 DataFrame 반환 + CSV 저장.
    get_news_items(soup)는 사용자가 만든(혹은 개선한) 함수를 그대로 호출한다고 가정.
    """
    dfs = []

    with requests.Session() as session:
        session.headers.update(HEADERS)

        for i in range(days):
            date_obj = datetime.date.today() - datetime.timedelta(days=i)
            date_str = date_obj.strftime("%Y-%m-%d")

            url = f"https://finance.naver.com/news/mainnews.naver?date={date_str}"
            r = session.get(url, timeout=timeout)
            r.raise_for_status()

            soup = BeautifulSoup(r.text, "lxml")

            # 해당일 기사 수집
            temp_df = get_news_items(soup)
            temp_df["date"] = date_obj  # 또는 date_str 로 저장하고 싶으면 date_str

            dfs.append(temp_df)

            if verbose:
                print(f"[{date_str}] {len(temp_df)} articles")
                print(temp_df.head(3))

            if sleep_sec:
                time.sleep(sleep_sec)

    news_df = pd.concat(dfs, ignore_index=True) if dfs else pd.DataFrame(columns=["기사제목", "본문url", "기사본문", "date"])

    # 중복 제거 (원하면 조정)
    if "본문url" in news_df.columns:
        news_df = news_df.drop_duplicates(subset=["본문url"], keep="first")

    news_df.to_csv(out_csv, index=False, encoding="utf-8-sig")
    return news_df
```
