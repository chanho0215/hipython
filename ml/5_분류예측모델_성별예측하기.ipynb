## 7.  위 4가지 모델의 학습 & 예측 & 평가 결과를 확인하고 최고 성능을 내는 모델을 찾아봅시다!

- 어떤 모델이 가장 성능이 좋은가요 ?

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

def get_clf_eval(y_test, pred):
    cm = confusion_matrix(y_test, pred)
    acc = accuracy_score(y_test, pred)
    prec = precision_score(y_test, pred, zero_division=0)
    rec = recall_score(y_test, pred, zero_division=0)
    f1 = f1_score(y_test, pred, zero_division=0)

    print("Confusion Matrix")
    print(cm)
    print(f"Accuracy : {acc:.4f}")
    print(f"Precision: {prec:.4f}")
    print(f"Recall   : {rec:.4f}")
    print(f"F1-score : {f1:.4f}")

    return acc, prec, rec, f1
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

results = []

# 1) Logistic Regression (표준화)
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train)
X_test_s = scaler.transform(X_test)

lr = LogisticRegression(max_iter=2000, solver='liblinear', random_state=0)
lr.fit(X_train_s, y_train)
lr_pred = lr.predict(X_test_s)

print("\n📌 LogisticRegression")
acc, prec, rec, f1 = get_clf_eval(y_test, lr_pred)
results.append(["LogisticRegression", acc, prec, rec, f1])


# 2) Decision Tree (원본)
dt = DecisionTreeClassifier(
    max_depth=4,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=0
)
dt.fit(X_train, y_train)
dt_pred = dt.predict(X_test)

print("\n📌 DecisionTreeClassifier")
acc, prec, rec, f1 = get_clf_eval(y_test, dt_pred)
results.append(["DecisionTreeClassifier", acc, prec, rec, f1])


# 3) Random Forest (원본)
rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=6,
    min_samples_split=10,
    min_samples_leaf=5,
    random_state=0,
    n_jobs=-1
)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

print("\n📌 RandomForestClassifier")
acc, prec, rec, f1 = get_clf_eval(y_test, rf_pred)
results.append(["RandomForestClassifier", acc, prec, rec, f1])


# 4) XGBoost (원본)
xgb = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1.0,
    random_state=0,
    eval_metric='logloss'
)
xgb.fit(X_train, y_train)
xgb_pred = xgb.predict(X_test)

print("\n📌 XGBClassifier")
acc, prec, rec, f1 = get_clf_eval(y_test, xgb_pred)
results.append(["XGBClassifier", acc, prec, rec, f1])


# 결과 표로 정리
result_df = pd.DataFrame(results, columns=["Model", "Accuracy", "Precision", "Recall", "F1"])
display(result_df.sort_values("F1", ascending=False))

best_model = result_df.sort_values("F1", ascending=False).iloc[0]
print(f"\n🏆 최고 성능 모델(F1 기준): {best_model['Model']}  |  F1 = {best_model['F1']:.4f}")
```

네 가지 분류 모델을 비교한 결과, 전반적으로 Accuracy는 0.65~0.69 수준으로 유사하게 나타났으나,<br>
Recall 및 F1-score에서는 차이가 관찰되었다.<br>
특히 XGBoost 모델이 F1-score 기준 가장 우수한 성능을 보였으며,<br>
이는 앙상블 기반 모델이 복잡한 비선형 패턴을 상대적으로 잘 포착했기 때문으로 해석된다.
